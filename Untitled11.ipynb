```python
import random
import time
from collections import deque, defaultdict
from colorama import Fore, Style
```

```python
id_klientow = []
for i in range(1,11):
  id_klientow.append(f'ACC{i:02d}')

miasta = ['Kraków', 'Warszawa', 'Lublin', 'Ryki', 'Dęblin', 'Puławy', 'Wrocław', 'Gołąb', 'Sandomierz', 'Radom']

def stworz_transakcje():
#client_id, kwota, miasto oraz aktualny timestamp
  transakcja = {
      'id_klienta': random.choice(id_klientow),
      'kwota': round(random.uniform(100, 100000), 2),
      'miasto': random.choice(miasta),
      'znacznik_czasu': time.time()
  }

  return transakcja

tran = stworz_transakcje()
print(tran)
```

```python
transakcje = defaultdict(lambda: deque(maxlen=5))
```

```python
# czy transakcja jest ok, czy nie
def analityczna(nowa_transakcja, historia_klienta):

  # nie możliwa podróż

  if len(historia_klienta) > 0:

    ostatnia = historia_klienta[-1]

    roznica_czasu = nowa_transakcja['znacznik_czasu'] - ostatnia['znacznik_czasu']

    inne_miasto = nowa_transakcja['miasto'] != ostatnia['miasto']

    if (roznica_czasu < 5 and inne_miasto):
      print(f"{Fore.RED}[ALERT]{Style.RESET_ALL} Niemożliwa podróż")
      return False

  # zbyt wiele transakcji

  teraz = nowa_transakcja['znacznik_czasu']

  szybkie_transakcje = sum(1 for t in historia_klienta if (teraz - t['znacznik_czasu']) < 5)

  if szybkie_transakcje >= 3:
    print(f"{Fore.RED}[ALERT]{Style.RESET_ALL} Za dużo transakcji w krótkim czasie")

    return False

  return True

```

```python
while (True):
  t = stworz_transakcje()

  klient = t['id_klienta']

  print("========================================================================================")
  print(f"NOWA TRANSAKCJA  | Klient: {klient} | Kwota: {t['kwota']}PLN | Miasto: {t['miasto']}")
  print("----------------------------------------------------------------------------------------")

  if (analityczna(t, transakcje[klient])):
    # ok
    print(f"Weryfikacja: [{Fore.GREEN}OK{Style.RESET_ALL}]")
  else:
    # nie ok
    print(f"[{Fore.RED}ALERT{Style.RESET_ALL}]Zablokowano podejrzana transakcję")

  print("========================================================================================")
  transakcje[klient].append(t)
  print("\nHistoria klienta:")
  licznik = 1
  for transakcja in transakcje[klient]:
    print(f"TRANSAKCJA{licznik} | Klient: {transakcja['id_klienta']} | Kwota: {transakcja['kwota']}PLN | Miasto: {transakcja['miasto']}")
    licznik += 1
  print("\n\n\n\n\n")
  time.sleep(random.uniform(0.2, 0.8))
```
